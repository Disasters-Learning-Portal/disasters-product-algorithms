```python
metadata, tifs = iceye_v2.retrieve_iceye_resources(datetime(2026, 7, 21, 20, 51, 37))
```

```python
saved_products = iceye_v2.sigmaCalib(tifs, metadata)
```

```python
transformed_products = geotools.transform_geotifs_to_projection(saved_products, saved_products.replace(".tif", "_wgs_84.tif"), 'EPSG:4326')
```

```python
no_data = [0, 0]

from shared_utils.plotting import preview_cogs
from shared_utils.cog_utils import convert_to_cog

for file_id in range(len(transformed_products)):
    if transformed_products[file_id] is not None:
        print(f"Cogging and previewing {transformed_products[file_id]}")
        cog_path = convert_to_cog(
            transformed_products[file_id],
            nodata=no_data[file_id],
            compression="ZSTD",
            compression_level=1,
            dst_crs = None # Note that these files are not in WGS 84, but we're not converting them here.
        )
        print(cog_path)
        preview_cogs([cog_path], sample_n=4)
```
