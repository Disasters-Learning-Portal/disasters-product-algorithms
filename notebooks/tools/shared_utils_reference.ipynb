## 1. SimpleProcessor

The main high-level interface for processing disaster imagery from S3.

````python
from shared_utils.notebook_helpers import SimpleProcessor

config = {
    'event_name': '202510_Flood_AK',
    'bucket': 'nasa-disasters',
    'source_path': 'drcs_activations/202510_Flood_AK/sentinel2',
    'destination_base': 'drcs_activations_new',

    # Default is 'EPSG:3857' (Web Mercator). None keeps the source projection.
    # See section 3 — do NOT use EPSG:4326 for anything headed to build_stac.
    'target_crs': 'EPSG:3857',
    'overwrite': False,
    'verify': True,
    'compression': 'ZSTD',
    'compression_level': 9,

    # Optional: categorization patterns, as {category_name: regex}.
    # NOTE the direction. SimpleProcessor takes name -> regex; the standalone
    # `file_naming.categorize_file` in section 7 takes the TRANSPOSE
    # (regex -> subdir). SimpleProcessor inverts this dict internally before
    # calling it. The three dicts below are all keyed by the category NAME.
    #
    # Supplying this REPLACES the built-in defaults (it does not merge with
    # them), so list every category you want.
    'categorization_patterns': {
        'trueColor': r'trueColor|truecolor|true_color|RGB',
        'colorInfrared': r'colorInfrared|colorIR|color_infrared',
        'NDVI': r'NDVI|ndvi',
    },

    # Optional: output subdirectory per category (relative to destination_base)
    'output_dirs': {
        'trueColor': 'Sentinel-2/trueColor',
        'colorInfrared': 'Sentinel-2/colorIR',
        'NDVI': 'Sentinel-2/NDVI',
    },

    # Optional: per-category nodata values
    'nodata_values': {
        'NDVI': -9999,
        'trueColor': None,  # auto-detect
    },
}

processor = SimpleProcessor(config)
````

````python
# Step 1: Connect to S3
processor.connect_to_s3()

# Step 2: Discover and categorize files
num_files = processor.discover_files()
print(f"Found {num_files} files")

# Limit to first 2 files per category (reference notebook only)
for category in processor.files_to_process:
    processor.files_to_process[category] = processor.files_to_process[category][:2]
total = sum(len(f) for f in processor.files_to_process.values())
print(f"Limited to {total} files for demo")

# Step 3: Preview what will be processed
processor.preview_processing()

# Step 4: Process all files
results_df = processor.process_all()
results_df.head()
````

## 3. target_crs

Control whether files are reprojected during COG conversion.

The library default is **EPSG:3857** (Web Mercator) at every level —
`cog_utils.convert_to_cog(dst_crs=...)`, `main_processor.convert_to_cog(target_crs=...)`,
and `SimpleProcessor`. Set `target_crs` to skip reprojection or pick a different CRS.

> **Do not output EPSG:4326.** PROJ writes the WGS 84 *ensemble* plus a lat-first
> axis order, which `rio_stac.get_dataset_geom` cannot handle — it raises
> `Point outside of projection domain` in `veda-data-airflow`'s `build_stac`.
> Web Mercator dodges both problems. See "Critical Constraints" in `CLAUDE.md`.

**Which to pick:**

| Value | When |
|---|---|
| `'EPSG:3857'` | Headed for the VEDA dashboard / `build_stac`. Optimal for titiler-pgstac's default `WebMercatorQuad` tiling. |
| `None` | Fastest (no warp). Fine for browser preview and `leafmap`; will fail in airflow until reprojected. This is what the operator notebook templates default to. |
| `'EPSG:326xx'` | A specific UTM zone, when a downstream consumer needs it. |

````python
_base = {
    'event_name': '202510_Flood_AK',
    'bucket': 'nasa-disasters',
    'source_path': 'drcs_activations/202510_Flood_AK/sentinel2',
    'destination_base': 'drcs_activations_new',
}

# Option 1: Web Mercator (library default). Use for anything going to build_stac.
config_webmerc = {**_base, 'target_crs': 'EPSG:3857'}

# Option 2: Keep the source CRS — no warp, fastest.
# Any of these mean "no reprojection": None, 'None', 'none', ''
config_native = {**_base, 'target_crs': None}

# Option 3: A specific projection, e.g. UTM Zone 6N (Alaska).
config_utm = {**_base, 'target_crs': 'EPSG:32606'}

# _normalize_target_crs is what SimpleProcessor applies to the config value.
for label, cfg in [('webmerc', config_webmerc), ('native', config_native), ('utm', config_utm)]:
    print(f"{label:8} -> {SimpleProcessor(cfg)._normalize_target_crs()}")
````

The `target_crs` parameter works at every level:

```python
# High-level: SimpleProcessor / quick_process
config = {'target_crs': None, ...}

# Mid-level: main_processor.convert_to_cog
convert_to_cog(..., target_crs=None)

# Low-level: GDAL processor
create_cog_gdal(..., target_crs=None)
process_file_optimized(..., target_crs=None)

# Local conversion: cog_utils.convert_to_cog
convert_to_cog(input_tif, dst_crs=None)
```

## 4. Local COG Conversion

Convert local GeoTIFF files to COG without S3.

````python
from shared_utils.cog_utils import convert_to_cog, validate_cog

# Convert a local file to COG.
output_path = convert_to_cog(
    input_tif='/path/to/input.tif',
    output_cog='/path/to/output_cog.tif',   # None replaces the input in place
    nodata=-9999,                # see the three-way contract below
    dst_crs='EPSG:3857',         # default; None keeps the source projection
    compression='ZSTD',
    compression_level=22,        # library default (templates use 9)
    overview_levels=5,
    backend='rio',               # 'rio' (rio-cogeo) or 'gdal'
    metadata=None,               # dict of activation tags -> embedded at creation
)
print(f"COG created: {output_path}")

# validate_cog returns (is_valid, details); details has 'valid'/'errors'/'warnings'.
is_valid, details = validate_cog(output_path)
print(f"Valid COG: {is_valid}")
print(details)

# `nodata` is a THREE-way contract, resolved by TYPE (never truthiness — 0 == False):
#   None      -> inherit the source's tag, else auto-detect from dtype
#   <number>  -> declare exactly that, after validate_nodata_for_dtype
#   False     -> declare NO nodata, and strip any tag the source carries
#                (for inputs that carry their own alpha/mask band)
#   True      -> ValueError (bool subclasses int; an unguarded True would
#                validate as the numeric 1 and be declared as the sentinel)
````

````python
# Auto-select nodata based on data type
from shared_utils.cog_utils import set_nodata_value, validate_nodata_for_dtype

print(set_nodata_value('float32'))   # -9999.0
print(set_nodata_value('uint8'))     # 0
print(set_nodata_value('int16'))     # -9999
print(set_nodata_value('int8'))      # -128

# An unsupported dtype raises rather than guessing:
try:
    set_nodata_value('complex64')
except ValueError as e:
    print(f"ValueError: {e}")

# Validate a nodata choice. Returns {'valid': bool, 'error': str | None}.
print(validate_nodata_for_dtype(-9999, 'uint8'))
# -> {'valid': False, 'error': 'Value -9999.0 out of range for uint8 [0, 255]'}
print(validate_nodata_for_dtype(0, 'uint8'))
# -> {'valid': True, 'error': None}
````

## 5. GeoTIFF Analysis

Analyze files before processing to understand their properties.

````python
from shared_utils.geotiff_analyzer import analyze_geotiff, format_analysis_report

# Analyze a local file
results = analyze_geotiff('/path/to/file.tif')   # optional sample_size=N for big files
print(format_analysis_report(results))

# Keys actually present in `results`:
#   results['file']             - path as passed in
#   results['file_size_mb']     - size on disk
#   results['crs']              - Coordinate Reference System (str)
#   results['width']            - pixels
#   results['height']           - pixels
#   results['band_count']       - number of bands
#   results['dtype']            - data type of band 1
#   results['nodata_current']   - nodata tag on the file, or None
#   results['bands']            - list of per-band dicts; stats live at
#                                 band['statistics']['min' | 'max' | ...]
#   results['suggested_nodata'] - dict, see the next cell
#   results['nodata_conflicts'] - only when nodata_current is not None
#
# There is no 'bounds', 'shape', 'nodata', or 'compression' key. For bounds or
# compression, open the file directly with rasterio, or use compare_geotiffs
# (section 9), whose input_metadata/output_metadata dicts do carry 'compression'.
````

````python
from shared_utils.geotiff_analyzer import suggest_nodata_value

# analyze_geotiff already ran this — results['suggested_nodata'] is the same dict.
# Calling it directly, note the arg is `current_nodata` and the file's own tag
# lives under 'nodata_current' (not 'nodata').
suggestion = suggest_nodata_value(
    dtype=results['dtype'],
    bands=results['bands'],
    current_nodata=results['nodata_current'],
)

# Returns {'value', 'reasoning', 'alternatives'} — note: NOT 'suggested_value'/'reason'.
print(f"Suggested nodata: {suggestion['value']}")
print(f"Reasoning:        {suggestion['reasoning']}")
print(f"Alternatives:     {suggestion['alternatives']}")
````

## 7. File Naming

Parse filenames, extract dates, generate standardized names.

````python
from shared_utils.file_naming import (
    extract_datetime_from_filename,
    categorize_file,
    create_output_filename,
)

# Extract the first datetime substring in a filename.
# Returns (matched_string, granularity) where granularity is 'hour' or 'day'.
matched, granularity = extract_datetime_from_filename('S2_trueColor_20251015_alaska.tif')
print(f"Matched: {matched}  granularity={granularity}")     # '20251015', 'day'

# ISO timestamps are recognized at hour granularity.
matched, granularity = extract_datetime_from_filename('sentinel_2025-10-15T19:46:16Z_red.tif')
print(f"Matched: {matched}  granularity={granularity}")     # '2025-10-15T19:46:16Z', 'hour'

# categorize_file takes {regex -> S3 subdir} and returns the SUBDIR (the dict
# VALUE), or 'uncategorized'. Case-insensitive, first match wins.
#
# ⚠️ This is the TRANSPOSE of SimpleProcessor's 'categorization_patterns'
# config in section 1, which is {category_name -> regex}. SimpleProcessor
# inverts its config before calling this. Passing a {name: regex} dict here
# does not raise — the name matches as a regex — it just hands back the regex
# string as the "category", which is exactly the bug that silently mis-routed
# SimpleProcessor's output. Check which direction you need.
CATEGORIES = {
    r'trueColor|truecolor|true_color': 'Sentinel-2/trueColor',
    r'colorInfrared|colorIR':          'Sentinel-2/colorIR',
}
print(f"Category: {categorize_file('S2_trueColor_20251015_alaska.tif', CATEGORIES)}")
print(f"Category: {categorize_file('S2_mystery.tif', CATEGORIES)}")   # 'uncategorized'

# Build a standardized output filename. Bare YYYYMMDD dates are auto-hyphenated
# to YYYY-MM-DD for legacy operator-facing compatibility.
cog_name = create_output_filename(
    'S2_trueColor_20251015_alaska.tif',
    '202510_Flood_AK',
)
print(f"COG name: {cog_name}")
# -> 202510_Flood_AK_S2_trueColor_alaska_2025-10-15_day.tif

# It is IDEMPOTENT: a stem already carrying the event prefix and/or a trailing
# canonical stamp is returned unchanged, so re-running over your own output is safe.
print(create_output_filename(cog_name, '202510_Flood_AK') == cog_name)   # True

# Compose the destination S3 key with a plain os.path.join — there is no
# longer a dedicated helper, since `os.path.join` already covers it.
import os
path = os.path.join('drcs_activations_new', 'Sentinel-2/trueColor', cog_name)
print(f"Output path: {path}")
````

## 8. Batch Analysis

Analyze multiple files in parallel and generate reports.

````python
from shared_utils.batch_analyzer import (
    analyze_batch_local,
    analyze_batch_s3,
    generate_summary_statistics,
    create_detailed_report,
)

# Analyze local files
import glob
files = glob.glob('/path/to/geotiffs/*.tif')[:2]  # Limit to first 2
results = analyze_batch_local(files, max_workers=4)

# Or analyze from S3 (limited to 2 files for demo)
results = analyze_batch_s3(
    bucket='nasa-disasters',
    prefix='drcs_activations/202510_Flood_AK/sentinel2',
    suffix='.tif',
    max_workers=4,
    limit=2,  # analyze first 2 files only
)

# Summary statistics. Real keys: total_files, successful, failed, data_types,
# nodata_values, min_values, max_values, file_sizes_mb, recommendations, plus
# overall_min / overall_max / avg_file_size_mb when any band had valid data.
# (There is no 'dtypes' key and no 'crs_values' key.)
summary = generate_summary_statistics(results)
print(f"Total files: {summary['total_files']}")
print(f"Successful:  {summary['successful']}  Failed: {summary['failed']}")
print(f"Data types:  {summary['data_types']}")       # {'uint8': 3, ...}
print(f"Nodata seen: {summary['nodata_values']}")    # {'None': 2, '-9999.0': 1}
print(f"Suggestions: {summary['recommendations']}")  # {'uint8:0': 3, ...}
if 'overall_min' in summary:
    print(f"Range: {summary['overall_min']} .. {summary['overall_max']}")
    print(f"Avg size: {summary['avg_file_size_mb']:.1f} MB")

# Detailed DataFrame report. Columns: file, size_mb, width, height, bands,
# dtype, current_nodata, has_error, then per-band band_N_{min,max,mean,
# nodata_pct}, suggested_nodata, suggestion_reason, has_nodata_conflict.
# Note CRS is NOT in this report — per-file CRS comes from analyze_geotiff's
# results['crs'] (section 5).
df = create_detailed_report(results)
df.head()
````

## 9. Verification

Compare input and output files to verify processing integrity.

````python
from shared_utils.verification import compare_geotiffs, verify_s3_files

# Compare local files
comparison = compare_geotiffs(
    input_path='/path/to/original.tif',
    output_path='/path/to/cog_output.tif',
    band=1,
    sample_size=10000,  # sample for large files; None reads the full band
)

# Top-level keys: input_file, output_file, band, timestamp,
#                 input_metadata, output_metadata, statistics, verification
#
# Pass/fail lives under 'verification' — there are no top-level
# 'data_preserved' / 'crs_match' / 'nodata_match' keys.
v = comparison['verification']
print(f"Passed:   {v['passed']}  ({v['status']})")
print(f"Warnings: {v['warnings']}")
print(f"Errors:   {v['errors']}")

# CRS / nodata / compression are compared by reading the two metadata dicts:
im, om = comparison['input_metadata'], comparison['output_metadata']
print(f"CRS:         {im['crs']} -> {om['crs']}  (match={im['crs'] == om['crs']})")
print(f"Nodata:      {im['nodata']} -> {om['nodata']}  (match={im['nodata'] == om['nodata']})")
print(f"Compression: {im['compression']} -> {om['compression']}")
print(f"Tiled:       {om['is_tiled']} @ {om['blockxsize']}x{om['blockysize']}")

# Band statistics: statistics['input'|'output'] each carry min/max/mean/std/
# median/valid_pixels/nodata_pixels, plus 'differences' and 'correlation'.
print(f"Correlation: {comparison['statistics'].get('correlation')}")

# Compare S3 files (downloads to /tmp, writes plots + a JSON report to save_dir).
# Returns the same dict, plus 'plots' and 'results_file'.
result = verify_s3_files(
    input_bucket='nasa-disasters',
    input_key='drcs_activations/original.tif',
    output_bucket='nasa-disasters',
    output_key='drcs_activations_new/output_cog.tif',
    save_dir='/tmp/verification',
)
print(result['results_file'])
````
